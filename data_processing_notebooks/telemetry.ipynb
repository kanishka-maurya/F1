```python
import pandas as pd
```

```python
df_tele = pd.read_parquet(
    r'C:\Users\Asus\Desktop\Formula1\data\bronze\telemetry\all_seasons_telemetry.parquet')
df_tele.head(5)
```

```python
df_tele.shape
```

```python
df_tele.info()
```

```python
df_tele.rename(columns={"driver_id": "driver_ref"}, inplace=True)
```

- no **PRIMARY KEY**
- combination of season, round_number, driver_code and lap_number is **PRIMARY KEY**

```python
df_tele["tele_ref"] = df_tele["season"].astype(str).str[2:4] + "_" + df_tele["round_number"].astype(str) + "_" + df_tele["driver_ref"].astype(str) + "_" + df_tele["lap_number"].astype(str)
```

```python
df_tele["tele_ref"].head(5)
```

- handling race_date datatype

```python
df_tele["race_date"] = pd.to_datetime(df_tele["race_date"])
```

```python
df_tele.info()
```

- changing datatype of driver_number

```python
df_tele["driver_number"] = (df_tele["driver_number"]).astype(int)
```

```python
df_tele.columns
```

```python
df_tele = df_tele[['tele_ref', 'season', 'round_number', 'race_name', 'circuit_ref', 'race_date',
       'driver_ref', 'driver_number', 'lap_number', 'sample_count', 'avg_speed_kph',
       'max_speed_kph', 'min_speed_kph', 'avg_throttle_pct',
       'full_throttle_pct', 'heavy_braking_pct', 'drs_active_pct', 'avg_gear',
       'gear_changes', 'avg_rpm', 'max_rpm']]
```
