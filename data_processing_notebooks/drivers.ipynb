- handling null values in nationality and date_of_birth

```python
driver_data = {
    'Paul Aron':        ('1999-02-04', 'Estonian'),  
    'Dino Beganovic':   ('2004-01-19', 'Swedish'),
    'Luke Browning':    ('2002-01-31', 'British'),
    'Jak Crawford':     ('2005-05-02', 'American'),
    'Felipe Drugovich': ('2000-05-23', 'Brazilian'),
    'Alexander Dunne':  ('2005-11-11', 'Irish'),
    'Antonio Fuoco':    ('1996-05-20', 'Italian'),
    'Ryo Hirakawa':     ('1994-03-07', 'Japanese'),
    'Ayumu Iwasa':      ('2001-09-22', 'Japanese'),
    'Arthur Leclerc':   ('2000-10-14', 'Monégasque'),
    'Victor Martins':   ('2001-06-16', 'French'),
    "Patricio O'Ward":  ('1999-05-06', 'Mexican'),
    'Cian Shields':     ('2005-03-07', 'British'),
    'Frederik Vesti':   ('2002-01-13', 'Danish'),
}
```

```python
dob_map = df_drivers['driver_name'].map(lambda x: driver_data.get(x, (None, None))[0])
nat_map = df_drivers['driver_name'].map(lambda x: driver_data.get(x, (None, None))[1])

df_drivers['date_of_birth'] = df_drivers['date_of_birth'].fillna(dob_map)
df_drivers['nationality']   = df_drivers['nationality'].fillna(nat_map)
```

- changing date_of_birth datatype

```python
df_drivers["date_of_birth"] = pd.to_datetime(df_drivers["date_of_birth"])
```

- forename and surname columns are redundant now - dropping them

```python
df_drivers.drop(columns=["forename", "surname"], inplace=True)
```
