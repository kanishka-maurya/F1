```python
import pandas as pd
import fastf1
from fastf1.ergast import Ergast
```

```python
# Initialize Ergast API client with desired settings
ergast = Ergast(
    result_type='pandas',  # 'pandas'
    auto_cast=True,        # Automatic type conversion
    limit=1000             # Max results per request (default: 30, max: 1000)
)
```

# Drivers

```python
drivers_df = pd.DataFrame()

for season in range(2010, 2026):
    print(f"Season: {season}")

    # Get all drivers from a season
    drivers = ergast.get_driver_info(season=season)

    drivers = drivers[["driverId", "driverNumber", "driverCode", "givenName", "familyName", "dateOfBirth", "driverNationality"]]

    drivers_df = pd.concat([drivers_df, drivers], ignore_index=True)

    print(drivers.columns)

```

```python
import requests
import pandas as pd
from pathlib import Path

BASE_URL  = "https://api.jolpi.ca/ergast/f1"

# ── Fetch all pages ───────────────────────────────────────────────────────────
print("Fetching all F1 drivers...")

all_drivers = []
limit       = 100   # fetch 100 per page
offset      = 0

while True:
    url      = f"{BASE_URL}/drivers.json?limit={limit}&offset={offset}"
    response = requests.get(url, timeout=30)
    response.raise_for_status()
    data     = response.json()

    drivers = data["MRData"]["DriverTable"]["Drivers"]
    total   = int(data["MRData"]["total"])

    all_drivers.extend(drivers)
    print(f"  Fetched {len(all_drivers)}/{total} drivers...")

    offset += limit
    if offset >= total:
        break

print(f"Total drivers fetched: {len(all_drivers)}")

# ── Build DataFrame ───────────────────────────────────────────────────────────
records = []
for d in all_drivers:
    records.append({
        "driver_ref":    d.get("driverId"),
        "driver_code":   d.get("code"),
        "driver_number": d.get("permanentNumber"),
        "forename":      d.get("givenName"),
        "surname":       d.get("familyName"),
        "date_of_birth": d.get("dateOfBirth"),
        "nationality":   d.get("nationality"),
    })

df = pd.DataFrame(records)

# ── Clean ─────────────────────────────────────────────────────────────────────
df = df.drop_duplicates(subset=["driver_ref"])

```

```python
import pandas as pd
import requests
from pathlib import Path
from src.utils.logger import logging


def fetch_all_drivers(bronze_dir: Path) -> pd.DataFrame:
    """
    Fetch all F1 drivers from Ergast.
    Saves response to bronze, returns cleaned DataFrame.
    """
    save_path = bronze_dir / "entities" / "drivers.parquet"

    # Idempotency — skip if already fetched
    if save_path.exists():
        logger.info("Drivers already fetched, loading from disk")
        return pd.read_parquet(save_path)

    logger.info("Fetching all drivers from Ergast...")
    
    BASE_URL  = "https://api.jolpi.ca/ergast/f1"
    
    all_drivers = []
    limit       = 100   # fetch 100 per page
    offset      = 0

    while True:
        url      = f"{BASE_URL}/drivers.json?limit={limit}&offset={offset}"
        try:
            response = requests.get(url, timeout=30)
            response.raise_for_status()
            data     = response.json()
        except Exception as e:
            raise(e)

        drivers = data["MRData"]["DriverTable"]["Drivers"]
        total   = int(data["MRData"]["total"])

        all_drivers.extend(drivers)
        print(f"  Fetched {len(all_drivers)}/{total} drivers...")

        offset += limit
        if offset >= total:
            break

    print(f"Total drivers fetched: {len(all_drivers)}")

    # ── Build DataFrame ───────────────────────────────────────────────────────────
    records = []
    for d in all_drivers:
        records.append({
            "driver_ref":    d.get("driverId"),
            "driver_code":   d.get("code"),
            "driver_number": d.get("permanentNumber"),
            "forename":      d.get("givenName"),
            "surname":       d.get("familyName"),
            "date_of_birth": d.get("dateOfBirth"),
            "nationality":   d.get("nationality"),
        })

    df = pd.DataFrame(records)

    # Deduplicate on meaningful key only
    df = df.drop_duplicates(subset=["driver_ref"])

    # Save to bronze
    save_path.parent.mkdir(parents=True, exist_ok=True)
    df.to_parquet(save_path, index=False, compression="snappy")
    logger.info(f"Saved {len(df)} drivers → {save_path}")

    return df

from pathlib import Path

df = fetch_all_drivers(Path(r'C:\Users\Asus\Desktop\Formula1_exp\data\bronze'))
df.head()
```
