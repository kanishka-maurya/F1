# **SCHEDULE**

```python
df_schedule = pd.read_parquet(r"C:\Users\Asus\Desktop\Formula1\data\bronze\schedule\all_seasons_schedule.parquet")
df_schedule.head()
```

- creating **PRIMARY KEY**
- combination of season and round_number columns is unique

```python
df_schedule["schedule_ref"] = df_schedule["season"].astype(str).str[2:4] + "_" + df_schedule["round_number"].astype(str)
```

```python
len(df_schedule["schedule_ref"].unique())
```

```python
df_schedule["schedule_ref"].nunique()
```

- handling race_time and race_date columns
- combining both the columns meticulously

```python
df_schedule['race_datetime'] = pd.to_datetime(
    df_schedule['race_date'].astype(str) + ' ' + df_schedule['race_time'].astype(str),
    utc=True
)
```

```python
df_schedule["race_datetime"].head()
```

- handling fp2_date, fp3_date and quali_date columns

```python
df_schedule[df_schedule["fp2_date"].isna()].head()
```

```python
df_schedule['fp1_date'] = pd.to_datetime(df_schedule['fp1_date'], errors='coerce')
df_schedule['fp2_date'] = pd.to_datetime(df_schedule['fp2_date'], errors='coerce')
df_schedule['fp3_date'] = pd.to_datetime(df_schedule['fp3_date'], errors='coerce')
df_schedule['quali_date'] = pd.to_datetime(df_schedule['quali_date'], errors='coerce')
```

- removing sprint_date, race_date and race_time column

```python
df_schedule.drop(columns=["race_date", "race_time", "sprint_date"], inplace=True)
```

```python
df_schedule = df_schedule[["schedule_ref", "season", "round_number", "race_name", "race_datetime", "circuit_ref",
                           "fp1_date", "fp2_date", "fp3_date", "quali_date", "has_sprint"]]
```
